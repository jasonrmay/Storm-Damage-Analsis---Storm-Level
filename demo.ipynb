```python
from dataGetter import DataGetter
from dataProcessor import DataProcessor
from merger import Merge
import os

CENSUS_KEY = os.getenv("CENSUS_KEY")

# The available years: 
# A census API key is not required but it is recommended to avoid rate limits and thus slow download rates.
dataGetter = DataGetter(CENSUS_KEY=CENSUS_KEY, year=2023, download=True,
                         roni_raw_data_path=r"static_data\RONI_data\rawData.csv",
                         coastal_type_raw_data_path=r"static_data\Coastal_data_2010\coastal_counties_2010.csv")
dat_2023 = dataGetter.fetch_all()

# note, the data processor will change dat_2023 due to the mutability of dictionaries, but not the downloaded data.
dataProcessor = DataProcessor(year=2023, data_dict=dat_2023, data_path=r"./cleanedData/", download=True)
cleaned_data_dict = dataProcessor.process_all()

# the cleaned data dict will be emptied out after merging.
merger = Merge(cleaned_data_dict, year=2023, download=True, save_path=r"./merged_data/")
merged_data = merger.merge()
```
